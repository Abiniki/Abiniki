```python
import requests

def analyze_security_headers(url):
    if not url.startswith(('http://', 'https://')):
        url = 'https://' + url

    # Key HTTP security headers and their recommendations
    security_headers = {
        "Strict-Transport-Security": {
            "desc": "Enforces secure HTTPS connections.",
            "rec": "Add 'Strict-Transport-Security: max-age=31536000; includeSubDomains'"
        },
        "Content-Security-Policy": {
            "desc": "Prevents Cross-Site Scripting (XSS) and data injection attacks.",
            "rec": "Define a strict Content-Security-Policy (CSP) restricting trusted sources."
        },
        "X-Frame-Options": {
            "desc": "Protects against Clickjacking attacks.",
            "rec": "Add 'X-Frame-Options: DENY' or 'SAMEORIGIN'"
        },
        "X-Content-Type-Options": {
            "desc": "Prevents MIME-sniffing vulnerabilities.",
            "rec": "Add 'X-Content-Type-Options: nosniff'"
        },
        "Referrer-Policy": {
            "desc": "Controls how much referrer information is shared.",
            "rec": "Add 'Referrer-Policy: strict-origin-when-cross-origin'"
        },
        "Permissions-Policy": {
            "desc": "Controls browser features like camera, microphone, and geolocation.",
            "rec": "Add a Permissions-Policy header restricting unused browser APIs."
        }
    }

    print(f"🔍 Scanning: {url}\n" + "-" * 50)
    try:
        response = requests.get(url, timeout=10)
        headers = response.headers

        present = []
        missing = []

        for header, info in security_headers.items():
            if header in headers:
                present.append((header, headers[header]))
            else:
                missing.append((header, info))

        print("✅ PRESENT HEADERS:")
        if present:
            for h, val in present:
                print(f"  - {h}: {val}")
        else:
            print("  None found.")

        print("\n❌ MISSING HEADERS & RECOMMENDATIONS:")
        if missing:
            for h, info in missing:
                print(f"  - {h}")
                print(f"    Why: {info['desc']}")
                print(f"    Fix: {info['rec']}\n")
        else:
            print("  All key security headers are present!")

    except requests.exceptions.RequestException as e:
        print(f"⚠️ Error connecting to {url}: {e}")

# Run the analyzer
target_url = input("Enter website URL (e.g., example.com): ")
analyze_security_headers(target_url)
```
